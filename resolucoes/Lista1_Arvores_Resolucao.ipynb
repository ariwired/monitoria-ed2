## Código Base

### Classe No

Representa um nó genérico de uma árvore binária, com valores de referência para os filhos esquerdo e direito.

````python
class No:
    """
    Classe que representa um nó em uma árvore binária.

    Atributos:
        valor: O valor armazenado no nó.
        esquerda: Referência ao filho esquerdo (None se não existir).
        direita: Referência ao filho direito (None se não existir).
    """
    def __init__(self, valor):
        self.valor = valor
        self.esquerda = None
        self.direita = None

    def __repr__(self):
        """Representação técnica do nó para depuração."""
        return str(self.valor)
````

### Classe ArvoreBinaria

Classe principal que gerencia a estrutura da árvore binária, oferecendo operações básicas como verificação de vazio, cálculo de altura, contagem de nós e visualização textual.

````python
class ArvoreBinaria:
    """
    Implementação de uma árvore binária genérica.

    Métodos principais:
    - esta_vazia()
    - altura()
    - __len__()          → número de nós
    - __str__()          → visualização textual bonita
    - __repr__()         → representação para depuração
    """
    def __init__(self, raiz=None):
        """
        Inicializa a árvore binária.

        Args:
            raiz (No | None): Nó raiz da árvore (opcional)
        """
        self.raiz = raiz

    def esta_vazia(self):
        """
        Verifica se a árvore não contém nenhum nó.

        Returns:
            bool: True se a árvore estiver vazia, False caso contrário.

        Complexidade: O(1)
        """
        return self.raiz is None

    def altura(self):
        """
        Calcula a altura da árvore.

        Returns:
            int: Altura da árvore (-1 se vazia, 0 para árvore com apenas raiz).

        Complexidade: O(n), onde n é o número de nós.
        """
        return self._altura_recursiva(self.raiz)

    def _altura_recursiva(self, no):
        """Método auxiliar recursivo para cálculo da altura."""
        if no is None:
            return -1  # Altura de árvore vazia
        return 1 + max(self._altura_recursiva(no.esquerda),
                       self._altura_recursiva(no.direita))

    def __len__(self):
        """
        Retorna o número total de nós na árvore.

        Returns:
            int: Quantidade de nós na árvore.

        Complexidade: O(n)
        """
        return self._tamanho_recursivo(self.raiz)

    def _tamanho_recursivo(self, no):
        """Método auxiliar recursivo para contagem de nós."""
        if no is None:
            return 0
        return 1 + self._tamanho_recursivo(no.esquerda) + self._tamanho_recursivo(no.direita)

    def __repr__(self):
        """Representação técnica da árvore (útil para depuração)."""
        return f"{self.__class__.__name__}(raiz={repr(self.raiz)})"

    def __str__(self):
        """
        Gera uma representação visual textual da árvore (visualização em pré-ordem invertida).

        Returns:
            str: Representação da árvore em formato de linhas indentadas.

        Complexidade: O(n)
        """
        if self.esta_vazia():
            return "<árvore vazia>"

        linhas = []
        self._coletar_linhas(self.raiz, "", linhas)
        return "\n".join(linhas)

    def _coletar_linhas(self, no, prefixo, linhas):
        """
        Método auxiliar recursivo para construção da representação textual.
        Exibe a subárvore direita primeiro (visualização "deitada" à direita).
        """
        if no is not None:
            # Visita primeiro o filho direito (para visualização mais natural)
            self._coletar_linhas(no.direita, prefixo + "   ", linhas)
            # Adiciona o nó atual
            linhas.append(prefixo + str(no.valor))
            # Visita o filho esquerdo
            self._coletar_linhas(no.esquerda, prefixo + "   ", linhas)
````

## Questão 1

Implemente a função soma_niveis_pares(raiz) que retorna a soma de todos os valores localizados em níveis pares da árvore, considerando a raiz como nível 0 (portanto, nível par). Caso a árvore esteja vazia, retorne 0


### Raciocínio

A informação da dica diz que é o nível do nó, ou seja, quantas arestas o separam da raiz. Cada elemento da fila passa a ser o par (nó, nível). Ao processar um nó, somamos o valor se o nível for par. Os filhos entram na fila com nível + 1.

Em vez de remover da frente da lista (pop(0), que é O(n)), usamos um índice i que avança pela lista. Os elementos já lidos ficam lá, mas não são mais acessados

````python
def soma_niveis_pares(raiz):
    """
        Soma os valores de todos os nós localizados em níveis pares.
        A raiz está no nível 0, que é par.

        Args:
            raiz (No | None): Raiz da árvore (ou subárvore).

        Returns:
            int: Soma dos valores nos níveis 0, 2, 4, ... (0 se a árvore for vazia).

        Complexidade: O(n) em tempo, pois cada nó entra e sai da fila uma vez. O(n) em espaço, pois a lista guarda todos os nós visitados.
    """
    
    if raiz is None:
        return 0 
    #caso base: árvore vazia não tem valores para somar.

    soma = 0

    fila = [(raiz, 0)] #cada elemento é uma tupla (nó, nível). A raiz começa no nível 0.

    i = 0 #posição do próximo elemento a ser processado. Avançar este índice faz o papel de "remover da frente da fila".

    while i < len(fila):
        no, nivel = fila[i]
        i += 1

        #nível par: o valor entra na soma.
        #nível ímpar: o nó é visitado, mas o valor é ignorado.
        if nivel % 2 == 0:
            soma += no.valor

        #os filhos estão um nível abaixo do pai.
        #só entram na fila se existirem.
        if no.esquerda is not None:
            fila.append((no.esquerda, nivel + 1))
        if no.direita is not None:
            fila.append((no.direita, nivel + 1))

    return soma
````

````python
print(soma_niveis_pares(arvore.raiz)) # 16 (nível 0: 1 | nível 2: 4 + 5 + 6)
print(soma_niveis_pares(None)) # 0 (árvore vazia)
print(soma_niveis_pares(No(7))) # 7 (só a raiz)
````

|---|---|---|---|
| 1 | 0 | sim | 1 |
| 2 | 1 | não | 1 |
| 3 | 1 | não | 1 |
| 4 | 2 | sim | 5 |
| 5 | 2 | sim | 10 |
| 6 | 2 | sim | 16 |

## Questão 2
Implemente a função caminhos_com_soma(no, alvo) que retorna uma lista de
listas, onde cada lista interna contém os valores de um caminho da raiz até uma folha cuja soma
seja exatamente igual a alvo. Pode existir mais de um caminho válido.

### Raciocínio
O molde é o de caminho_ate_valor, com append ao entrar no nó e pop ao sair (backtracking). 

Há três diferenças em relação àquela função:

1. A soma só é conferida em folhas, pois o enunciado pede caminhos da raiz até uma folha.
2. Ao achar um caminho válido, guardamos uma cópia (caminho.copy()). A lista caminho continua sendo alterada depois. Sem a cópia, todas as respostas apontariam para a mesma lista, que termina vazia.
3. Achar um caminho não encerra a busca, porque pode haver outros

````python
def caminhos_com_soma(no, alvo):
    """
        Encontra todos os caminhos da raiz até uma folha cuja soma é igual a alvo.

        Args:
            no (No | None): Raiz da árvore (ou subárvore).
            alvo (int): Soma desejada.

        Returns:
            list[list]: Lista de caminhos; cada caminho é uma lista de valores, da raiz até a folha. Lista vazia se nenhum caminho servir.

        Complexidade: O(n * h) no pior caso. Cada nó é visitado uma vez, mas cada caminho encontrado é copiado (custo até h, a altura da árvore).
    """
    
    resultado = [] #acumula as listas-resposta
    caminho = [] #caminho atual, da raiz até o nó que está sendo visitado

    def dfs(atual, soma_parcial):
        if atual is None:
            return #caso base: passou do fim de um ramo, não há o que fazer.

        caminho.append(atual.valor)
        soma_parcial += atual.valor

        eh_folha = atual.esquerda is None and atual.direita is None

        if eh_folha:
            #fim de um caminho completo: confere a soma.
            if soma_parcial == alvo:
                #guarda uma cópia, pois a lista original ainda será modificada.
                resultado.append(caminho.copy())
        else:
            #nó interno: explora os dois lados. não há return ao achar uma resposta, então a busca continua.
            dfs(atual.esquerda, soma_parcial)
            dfs(atual.direita, soma_parcial)

        #desfaz: ao voltar para o pai, remove este nó do caminho para que o próximo ramo comece do estado correto (backtracking).
        caminho.pop()

    dfs(no, 0)
    return resultado
````

````python
#arvore A: caminhos até folhas somam 7 (1-2-4), 8 (1-2-5) e 10 (1-3-6)

print(caminhos_com_soma(arvore.raiz, 8)) # [[1, 2, 5]]
print(caminhos_com_soma(arvore.raiz, 10)) # [[1, 3, 6]]
print(caminhos_com_soma(arvore.raiz, 99)) # []
print(caminhos_com_soma(None, 0)) # []
print("---------------------------------------------------------------")

# Árvore B (mais de um caminho válido):
#
#            5
#           / \
#          4   8
#         /   / \
#       11   13  4
#       / \     / \
#      7   2   5   1


b = No(5)
b.esquerda = No(4)
b.esquerda.esquerda = No(11)
b.esquerda.esquerda.esquerda = No(7)
b.esquerda.esquerda.direita = No(2)
b.direita = No(8)
b.direita.esquerda = No(13)
b.direita.direita = No(4)
b.direita.direita.esquerda = No(5)
b.direita.direita.direita = No(1)

print(ArvoreBinaria(b))

print("---------------------------------------------------------------")
print(caminhos_com_soma(b, 22)) # [[5, 4, 11, 2], [5, 8, 4, 5]]
````

## Questão 3

Implemente a função arvore_espelhada(no) que retorna a raiz de uma nova árvore, espelho da original, sem modificar em nada os nós recebidos como entrada.

### Raciocínio

espelhar_arvore(no) trocava os ponteiros dos próprios nós. Aqui isso não é feito, então a regra é nunca atribuir nada a um nó recebido. Para cada nó de valor x:

1. Cria um nó novo No(x).
2. O filho esquerdo do novo é o espelho do filho direito do original.
3. O filho direito do novo é o espelho do filho esquerdo do original.

````python
def arvore_espelhada(no):
    """
        Constrói uma NOVA árvore que é o espelho da árvore recebida.
        A árvore original não é modificada e nenhum de seus nós é reaproveitado.

        Args:
            no (No | None): Raiz da árvore original (ou subárvore).

        Returns:
            No | None: Raiz da nova árvore espelhada (None se a original for vazia).

        Complexidade: O(n) em tempo e em espaço (n nós novos são criados), mais O(h) de pilha de recursão.
    """

    if no is None:
        return None #caso base: o espelho de uma árvore vazia é uma árvore vazia.

    novo = No(no.valor) #cria um nó NOVO com o mesmo valor. O original não é tocado.

    # Troca de lados:
    #   esquerda do novo  <- espelho da DIREITA do original
    #   direita do novo   <- espelho da ESQUERDA do original
    
    novo.esquerda = arvore_espelhada(no.direita)
    novo.direita = arvore_espelhada(no.esquerda)

    return novo
````

````python
espelho = ArvoreBinaria(arvore_espelhada(arvore.raiz))

print("Espelho:")
print(espelho)
print("Original:")
print(arvore)

print("---------------------------------------------------------------")
print()

#1) A original não mudou:
print(valores_pre_ordem(arvore.raiz)) # [1, 2, 4, 5, 3, 6]

#2) O espelho tem a estrutura invertida:
print(valores_pre_ordem(espelho.raiz)) # [1, 3, 6, 2, 5, 4]

#3) Nenhum nó foi reaproveitado (são objetos diferentes):
print(espelho.raiz is arvore.raiz) # False
print(espelho.raiz.esquerda is arvore.raiz.direita) # False

#4) Espelho do espelho volta à forma original:
de_volta = arvore_espelhada(espelho.raiz)
print(valores_pre_ordem(de_volta)) # [1, 2, 4, 5, 3, 6]

# 5) Árvore vazia:
print(arvore_espelhada(None)) # None
````

## Questão 4

Implemente a função menor_ancestral_comum(no, valor1, valor2) que retorna o valor do menor ancestral comum entre dois valores existentes na árvore.

### Raciocinio

A chamada recursiva devolve o nó que encontrou (ou None). Cada nó decide com base no que voltou das duas subárvores:

| esquerda devolveu | direita devolveu | o que fazer |
|---|---|---|
| um nó | um nó | os valores estão em lados diferentes, então o nó atual é o ancestral |
| um nó | None | repassa o que veio da esquerda para cima |
| None | um nó | repassa o que veio da direita para cima |
| None | None | nada encontrado aqui, devolve None |

Se o nó atual é um dos valores buscados, ele devolve a si mesmo sem descer mais. Isso trata o caso em que um valor é ancestral do outro: o segundo valor está abaixo do primeiro, e o primeiro encontrado já é a resposta.

````python
def menor_ancestral_comum(no, valor1, valor2):
    """
        Encontra o menor ancestral comum entre dois valores da árvore.

        É o nó mais profundo que tem os dois valores em sua subárvore (um nó conta como ancestral de si mesmo).

        Args:
            no (No | None): Raiz da árvore (ou subárvore).
            valor1: Primeiro valor (assume-se que existe na árvore).
            valor2: Segundo valor (assume-se que existe na árvore).

        Returns:
            O valor do menor ancestral comum, ou None se a árvore for vazia.

        Complexidade: O(n) em tempo (cada nó é visitado no máximo uma vez). O(h) em espaço, pela pilha de recursão.
    """

    def busca(atual):
        if atual is None:
            return None #caso base: ramo sem nós, nada foi encontrado aqui.

        #achou um dos valores: devolve ESTE nó. não precisa descer mais, porque se o outro valor estiver abaixo, este nó já é o ancestral.
        if atual.valor == valor1 or atual.valor == valor2:
            return atual 

        #pergunta às duas subárvores se elas contêm algum dos valores.
        esq = busca(atual.esquerda)
        dir = busca(atual.direita)

        #um valor em cada lado: este é o primeiro ponto onde os caminhos se encontram, portanto é o menor ancestral comum.
        if esq is not None and dir is not None:
            return atual

        #só um lado (ou nenhum) achou algo: repassa para cima.
        return esq if esq is not None else dir

    resultado = busca(no)
    return resultado.valor if resultado is not None else None
````

Execução para menor_ancestral_comum(arvore.raiz, 4, 5):

```
busca(1): não é 4 nem 5 → pergunta aos filhos
    busca(2): não é 4 nem 5 → pergunta aos filhos
        busca(4) → devolve o nó 4           (achou valor1)
        busca(5) → devolve o nó 5           (achou valor2)
        os dois lados devolveram algo → devolve o nó 2   ← ancestral comum
    
    busca(3): não é 4 nem 5 → pergunta aos filhos
        busca(None) → None
        busca(6)    → None                  (6 não é alvo)
        devolve None
    
    esq = nó 2, dir = None → repassa o nó 2

resultado = nó 2 → retorna 2
```

````python
print(menor_ancestral_comum(arvore.raiz, 4, 5)) # 2 (irmãos: o pai deles)
print(menor_ancestral_comum(arvore.raiz, 4, 6)) # 1 (lados diferentes da raiz)
print(menor_ancestral_comum(arvore.raiz, 5, 6)) # 1
print(menor_ancestral_comum(arvore.raiz, 2, 5)) # 2 (um é ancestral do outro)
print(menor_ancestral_comum(arvore.raiz, 3, 6)) # 3 (idem)
print(menor_ancestral_comum(arvore.raiz, 1, 6)) # 1 (a raiz é um dos valores)
print(menor_ancestral_comum(arvore.raiz, 4, 4)) # 4 (mesmo valor duas vezes)
print(menor_ancestral_comum(None, 1, 2)) # None
````
