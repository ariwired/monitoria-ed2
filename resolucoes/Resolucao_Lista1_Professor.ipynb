### Classe No
Representa um nó genérico de uma árvore binária, contendo um valor e referências para os filhos esquerdo e direito.

````python
class No:
    """
    Classe que representa um nó em uma árvore binária.

    Atributos:
        valor: O valor armazenado no nó.
        esquerda: Referência ao filho esquerdo (None se não existir).
        direita: Referência ao filho direito (None se não existir).
    """
    def __init__(self, valor):
        self.valor = valor
        self.esquerda = None
        self.direita = None

    def __repr__(self):
        """Representação técnica do nó para depuração."""
        return str(self.valor)
````

### Classe ArvoreBinaria
Classe principal que gerencia a estrutura da árvore binária, oferecendo operações básicas como verificação de vazio, cálculo de altura, contagem de nós e visualização textual.

````python
class ArvoreBinaria:
    """
    Implementação de uma árvore binária genérica.

    Métodos principais:
    - esta_vazia()
    - altura()
    - __len__()          → número de nós
    - __str__()          → visualização textual bonita
    - __repr__()         → representação para depuração
    """
    def __init__(self, raiz=None):
        """
        Inicializa a árvore binária.

        Args:
            raiz (No | None): Nó raiz da árvore (opcional)
        """
        self.raiz = raiz

    def esta_vazia(self):
        """
        Verifica se a árvore não contém nenhum nó.

        Returns:
            bool: True se a árvore estiver vazia, False caso contrário.

        Complexidade: O(1)
        """
        return self.raiz is None

    def altura(self):
        """
        Calcula a altura da árvore.

        Returns:
            int: Altura da árvore (-1 se vazia, 0 para árvore com apenas raiz).

        Complexidade: O(n), onde n é o número de nós.
        """
        return self._altura_recursiva(self.raiz)

    def _altura_recursiva(self, no):
        """Método auxiliar recursivo para cálculo da altura."""
        if no is None:
            return -1  # Altura de árvore vazia
        return 1 + max(self._altura_recursiva(no.esquerda),
                       self._altura_recursiva(no.direita))

    def __len__(self):
        """
        Retorna o número total de nós na árvore.

        Returns:
            int: Quantidade de nós na árvore.

        Complexidade: O(n)
        """
        return self._tamanho_recursivo(self.raiz)

    def _tamanho_recursivo(self, no):
        """Método auxiliar recursivo para contagem de nós."""
        if no is None:
            return 0
        return 1 + self._tamanho_recursivo(no.esquerda) + self._tamanho_recursivo(no.direita)

    def __repr__(self):
        """Representação técnica da árvore (útil para depuração)."""
        return f"{self.__class__.__name__}(raiz={repr(self.raiz)})"

    def __str__(self):
        """
        Gera uma representação visual textual da árvore (visualização em pré-ordem invertida).

        Returns:
            str: Representação da árvore em formato de linhas indentadas.

        Complexidade: O(n)
        """
        if self.esta_vazia():
            return "<árvore vazia>"

        linhas = []
        self._coletar_linhas(self.raiz, "", linhas)
        return "\n".join(linhas)

    def _coletar_linhas(self, no, prefixo, linhas):
        """
        Método auxiliar recursivo para construção da representação textual.
        Exibe a subárvore direita primeiro (visualização "deitada" à direita).
        """
        if no is not None:
            # Visita primeiro o filho direito (para visualização mais natural)
            self._coletar_linhas(no.direita, prefixo + "   ", linhas)
            # Adiciona o nó atual
            linhas.append(prefixo + str(no.valor))
            # Visita o filho esquerdo
            self._coletar_linhas(no.esquerda, prefixo + "   ", linhas)
````

# Questão 1 — Média dos valores da árvore

**Conteúdo relacionado:** Percursos em Árvores (DFS) e Acumuladores.

## Enunciado

Implemente o método `media(self)` na classe `ArvoreBinaria`, que retorna a média aritmética de todos os valores armazenados. Caso a árvore esteja vazia, retorne `None`.

## Fundamentação teórica

Esta questão aplica o conceito de **caminhamento em árvore**. Para calcular a média, é necessário visitar cada nó exatamente uma vez, resultando em complexidade de tempo $O(n)$.

A estratégia utiliza uma função recursiva que retorna duas informações:

- a **soma** dos valores da subárvore;
- a **quantidade** de nós da subárvore.

Dessa forma, cada chamada recursiva recebe as informações calculadas pelas subárvores esquerda e direita e as combina com o valor do nó atual. Isso demonstra como a recursão pode fazer informações **"subirem" das subárvores até a raiz**.

Ao final do percurso, a média é calculada por:

$$
\text{média} = \frac{\text{soma dos valores}}{\text{quantidade de nós}}
$$

Seguindo o padrão da classe, a lógica recursiva fica isolada em um método auxiliar privado `(_soma_qtd)`, enquanto `media()` permanece como uma interface pública simples, sem parâmetros.

````python
def media(self):
    """
      Calcula a média aritmética de todos os valores armazenados na árvore.

      Returns:
          float | None: A média dos valores, ou None se a árvore estiver vazia.

      Complexidade: O(n) em tempo, O(h) em memória (pilha de recursão).
    """
    if self.esta_vazia():
        return None
    soma, qtd = self._soma_qtd(self.raiz)
    return soma / qtd

def _soma_qtd(self, no):
    """Auxiliar recursivo: retorna (soma dos valores, quantidade de nós)."""
    if no is None:
        return 0, 0
    soma_esq, qtd_esq = self._soma_qtd(no.esquerda)
    soma_dir, qtd_dir = self._soma_qtd(no.direita)
    return no.valor + soma_esq + soma_dir, 1 + qtd_esq + qtd_dir

ArvoreBinaria.media = media
ArvoreBinaria._soma_qtd = _soma_qtd

#Aqui usamos uma técnica do Python chamada monkey patching.
#Como a classe ArvoreBinaria já foi definida acima, podemos anexar novos métodos a ela depois.
#Isso evita reescrever a classe inteira a cada questão, mantendo cada resolução isolada em sua própria célula.
````

````python
# vai descomentando para testar.

raiz = No(10) # média: 10.0
#raiz.esquerda = No(5) # média: (10+5)/2 = 7.5
#raiz.direita = No(15) # média: (10+5+15)/3 = 10.0
#raiz.esquerda.esquerda = No(3) # média: (10+5+15+3)/4 = 8.25
#raiz.esquerda.direita = No(7) # média: (10+5+15+3+7)/5 = 8.0

arvore = ArvoreBinaria(raiz)
print(arvore.media()) # média atual, conforme os nós descomentados acima

# com tudo descomentado: (10+5+15+3+7)/5 = 8.0

# caso extra: árvore vazia
arvore_vazia = ArvoreBinaria()
print(arvore_vazia.media()) # none
````

# Questão 2 — Caminho de Soma Máxima

**Conteúdo relacionado:** Busca em Profundidade (DFS) e Backtracking.

## Enunciado

Retorne uma lista com os valores do caminho da raiz até uma folha cuja soma seja a maior possível.

## Fundamentação teórica

Esta questão explora o conceito de **backtracking**. Diferente de apenas acumular uma soma, é preciso manter o rastro do caminho percorrido até o momento. Ao retroceder na recursão (`pop`), o estado é desfeito, permitindo explorar novas ramificações sem "vazar" valores de um caminho para outro.

A estratégia aqui evita o uso de variáveis externas mutáveis (`nonlocal`): em vez disso, cada chamada recursiva **retorna** a melhor soma e o melhor caminho encontrados em sua subárvore, e esses resultados sobem e são comparados nó a nó, a mesma ideia de **"informação subindo pela recursão"** já utilizada na Questão 1.

Seguindo o padrão da classe, a lógica recursiva fica isolada em um método auxiliar privado (`_dfs_caminho_maximo`), enquanto `caminho_soma_maxima()` permanece como a interface pública, sem parâmetros.

````python
def caminho_soma_maxima(self):
    """
      Retorna o caminho da raiz até uma folha cuja soma dos valores seja máxima.

      Returns:
          list: valores do caminho de soma máxima, ou lista vazia se a árvore estiver vazia.

      Complexidade: O(n²) no pior caso em tempo e O(n²) no pior caso em memória, devido às cópias dos caminhos realizadas durante a recursão.
    """
    if self.esta_vazia():
        return []

    _, melhor_caminho = self._dfs_caminho_maximo(self.raiz, [], 0)
    return melhor_caminho

def _dfs_caminho_maximo(self, no, caminho_atual, soma_atual):
    """
      Auxiliar recursivo que explora todos os caminhos raiz-folha via backtracking, retornando a maior soma encontrada e o caminho correspondente.

      Args:
          no (No | None): nó atual sendo visitado.
          caminho_atual (list): caminho acumulado até este nó (mutado e revertido via pop).
          soma_atual (int | float): soma acumulada até este nó.

      Returns:
          tuple[float, list]: (maior soma encontrada, melhor caminho encontrado).
    """
    if no is None:
        return float('-inf'), []

    caminho_atual.append(no.valor)
    soma_atual += no.valor

    if no.esquerda is None and no.direita is None:
        resultado = (soma_atual, caminho_atual.copy())
    else:
        soma_esq, caminho_esq = self._dfs_caminho_maximo(no.esquerda, caminho_atual, soma_atual)
        soma_dir, caminho_dir = self._dfs_caminho_maximo(no.direita, caminho_atual, soma_atual)
        resultado = (soma_esq, caminho_esq) if soma_esq >= soma_dir else (soma_dir, caminho_dir)

    caminho_atual.pop()

    return resultado

ArvoreBinaria.caminho_soma_maxima = caminho_soma_maxima
ArvoreBinaria._dfs_caminho_maximo = _dfs_caminho_maximo
````


# Questão 3: Esvaziar Árvore (Pós-Ordem)

**Conteúdo relacionado:** Percurso Pós-Ordem (LRN) e Recursão.

## Enunciado

Remova todos os nós da árvore. Os filhos devem ser processados antes dos pais.

## Fundamentação teórica

O uso do Percurso Pós-Ordem é fundamental nesta questão, pois os filhos devem ser processados antes do nó pai.

A implementação segue a ordem **Esquerda → Direita → Raiz**, percorrendo recursivamente cada subárvore. Ao finalizar o processamento de um nó, suas referências para os filhos são definidas como None.

Ao final do percurso, self.raiz = None é utilizado para deixar a árvore vazia.

Essa abordagem garante uma limpeza lógica consistente com a estrutura de dependência da árvore, pois cada nó só é finalizado depois que suas subárvores já foram processadas.

````python
def esvaziar_arvore(self):
    """
      Remove todos os nós da árvore, desconectando os filhos de cada nó antes de desconectar o próprio nó (percurso pós-ordem).

      O percurso pós-ordem garante que os filhos sejam processados antes
      do pai, respeitando a ordem de dependência da estrutura: só faz
      sentido "descartar" um nó depois que suas subárvores já foram
      tratadas.

      Returns:
          None

      Complexidade:
          Tempo: O(n), onde n é o número de nós. Cada nó é visitado exatamente uma vez.
          Espaço: O(h), onde h é a altura da árvore, devido à pilha de recursão. Pior caso O(n) em árvore degenerada.
    """
    self._pos_ordem_limpar(self.raiz)
    self.raiz = None


def _pos_ordem_limpar(self, no):
    """
      Método auxiliar recursivo que percorre a subárvore em pós-ordem,
      desconectando os filhos de cada nó visitado.

      Args:
          no: Nó atual da subárvore sendo esvaziada.

      Returns:
          None

      Complexidade:
          Tempo: O(n) somando todas as chamadas.
          Espaço: O(h) pela pilha de recursão.
    """
    if no is None:
        return

    self._pos_ordem_limpar(no.esquerda)
    self._pos_ordem_limpar(no.direita)
    no.esquerda = None
    no.direita = None

ArvoreBinaria.esvaziar_arvore = esvaziar_arvore
ArvoreBinaria._pos_ordem_limpar = _pos_ordem_limpar
````

````python
arvore = ArvoreBinaria()

arvore.raiz = No(50)

arvore.raiz.esquerda = No(30)
arvore.raiz.direita = No(70)

arvore.raiz.esquerda.esquerda = No(20)
arvore.raiz.esquerda.direita = No(40)

arvore.raiz.direita.esquerda = No(60)
arvore.raiz.direita.direita = No(80)

arvore.raiz.esquerda.esquerda.esquerda = No(10)
arvore.raiz.direita.direita.direita = No(90)

print(f"Antes de esvaziar: raiz era {arvore.raiz.valor}")
print('')

arvore.esvaziar_arvore()

print(f"Depois de esvaziar: raiz virou {arvore.raiz}")
print('')
print(f"Está vazia? {arvore.esta_vazia()}")
````

# Questão 4: Árvore Simétrica (Espelho)

**Conteúdo relacionado:** Recursão e Propriedades Estruturais de Árvores.

## Enunciado

Verifique se a árvore é um espelho de si mesma.

## Fundamentação teórica

Uma árvore é simétrica quando sua subárvore esquerda é a imagem espelhada da subárvore direita.

A comparação não depende apenas dos valores dos nós, mas também de suas posições. Assim, o filho esquerdo de uma subárvore deve corresponder ao filho direito da outra, e vice-versa.

Diferente das questões anteriores, a recursão avança simultaneamente sobre duas subárvores. Por isso, os casos-base precisam considerar:

- ambos os nós iguais a None;
- apenas um dos nós igual a None;
- ambos existentes, permitindo comparar seus valores e continuar a recursão.

Na implementação utilizada, eh_simetrica() funciona como método público, enquanto _espelho() concentra a lógica recursiva da comparação.

As chamadas recursivas são feitas de forma cruzada:

- `no1.esquerda` com `no2.direita`;
- `no1.direita` com `no2.esquerda`.

Isso garante que a estrutura seja comparada como um espelho.

````python
def eh_simetrica(self):
    """
      Verifica se a árvore é um espelho de si mesma.

      Returns:
          bool: True se a árvore for simétrica, False caso contrário.

      Complexidade:
          Tempo: O(n)
          Espaço: O(h)
    """
    if self.esta_vazia():
        return True

    return self._espelho(self.raiz.esquerda, self.raiz.direita)


def _espelho(self, no1, no2):
    """
      Verifica recursivamente se duas subárvores são espelhadas entre si.
    """
    if no1 is None and no2 is None:
        return True

    if no1 is None or no2 is None:
        return False

    return (
        no1.valor == no2.valor
        and self._espelho(no1.esquerda, no2.direita)
        and self._espelho(no1.direita, no2.esquerda)
    )


ArvoreBinaria.eh_simetrica = eh_simetrica
ArvoreBinaria._espelho = _espelho
````

````python
arvore = ArvoreBinaria()
arvore.raiz = No(1)

print("Apenas a raiz:", arvore.eh_simetrica())  # True

arvore.raiz.esquerda = No(2)
arvore.raiz.direita = No(2)

print("Dois filhos iguais:", arvore.eh_simetrica())  # True


# Descomente as linhas abaixo para completar a árvore:

#arvore.raiz.esquerda.esquerda = No(3)
#arvore.raiz.esquerda.direita = No(4)
#arvore.raiz.direita.esquerda = No(4)
#arvore.raiz.direita.direita = No(3)

#print("Árvore completa:", arvore.eh_simetrica())  # True


# Depois, quebre a simetria descomentando:

#arvore.raiz.direita.direita.valor = 99

#print("Após alterar 3 para 99:", arvore.eh_simetrica())  # False
````
